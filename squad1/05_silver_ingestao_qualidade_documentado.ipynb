### Regras/KPIs gerados por este notebook:

Cada tabela tem 10 regras. As regras **por linha** viram colunas booleanas na Silver (`True` = passou, `False` = falhou, `null` = não avaliada). As regras **agregadas por pedido** não existem por linha (dependem de soma, de ausência ou de histórico), então são calculadas e gravadas no `dq_monitoring_logs` pelo notebook **06**.

**`ecommerce_itens_pedido`** (8 regras por linha + 2 agregadas):
- Técnica 1 — `tecnica_1_pk_valida`: `id_item_pedido` não nulo e não duplicado no micro-lote
- Técnica 2 — `tecnica_2_fk_pedido_valido`: `id_pedido` existe em `ecommerce_pedidos` (FK)
- Técnica 3 — `tecnica_3_fk_sku_valido`: `sku` existe em `ecommerce_produtos` (FK)
- Técnica 4 — `tecnica_4_quantidade_valida`: `quantidade` é inteiro e >= 1
- Técnica 5 — `tecnica_5_preco_valido`: `preco_unitario` > 0
- Negócio 6 — `negocio_6_desconto_nao_maior_preco`: `desconto_aplicado` <= `preco_unitario`
- Negócio 7 (agregada, notebook 06): soma dos itens == `valor_total` do pedido (tolerância R$ 0,01)
- Negócio 8 — `negocio_8_desconto_nao_negativo`: `desconto_aplicado` >= 0
- Negócio 9 — `negocio_9_desconto_dentro_limite`: `desconto_aplicado` <= 50% do `preco_unitario`
- Negócio 10 (agregada, notebook 06): todo pedido tem pelo menos 1 item

**`ecommerce_rastreamento_entregas`** (6 regras por linha + 4 agregadas):
- Técnica 1 — `tecnica_1_pk_valida`: `id_rastreamento` não nulo e não duplicado no micro-lote
- Técnica 2 — `tecnica_2_status_valido`: `status_entrega` dentro da lista de status válidos
- Técnica 3 — `tecnica_3_fk_pedido_valido`: `id_pedido_ecommerce` existe em `ecommerce_pedidos` (FK)
- Técnica 4 — `tecnica_4_data_valida`: `dt_evento` não nulo e não futuro
- Técnica 5 — `tecnica_5_codigo_rastreio_valido`: `codigo_rastreio` no formato `TRK######`
- Negócio 6 — `negocio_6_ordem_cronologica`: eventos do mesmo pedido em ordem cronológica (considera o histórico já gravado)
- Negócio 7 (agregada, notebook 06): pedido com status "Entregue" precisa ter evento "entregue"
- Negócio 8 (agregada, notebook 06): prazo entre "coletado" e "entregue" <= 30 dias
- Negócio 9 (agregada, notebook 06): pedido cancelado não pode ter evento de rastreamento
- Negócio 10 (agregada, notebook 06): pedido não cancelado deve ter uma única transportadora

As FKs que dependem da Silver de `ecommerce_pedidos` (de outra dupla) ficam **pendentes** (`null`) quando essa Silver ainda não está disponível, em vez de falhar — ver Etapa 5.

### Etapa 4 — Funções auxiliares (existência de tabela, referências e detecção incremental)

`tabela_delta_existe` evita erro ao checar uma tabela que pode não existir ainda; `carregar_referencias` carrega as dependências de `pedidos`/`produtos` (ficou como função auxiliar — o loop principal, mais abaixo, recarrega essas referências inline a cada verificação em vez de chamar essa função, pra sempre pegar dado atualizado); `detectar_novo_bronze` filtra, via log de controle, só os arquivos Bronze que ainda não foram promovidos pra Silver.


```python
def tabela_delta_existe(caminho):
    try:
        spark.read.format("delta").options(**adls_options).load(caminho).limit(1).count()
        return True
    except Exception:
        return False

print("Função de verificação de tabela Delta pronta.")
```

```python
def carregar_referencias():
    df_silver_pedidos = (spark.read.format("delta").options(**adls_options)
        .load(caminhos["silver_pedidos"])
        .select("id_pedido").distinct())

    df_silver_produtos = (spark.read.format("delta").options(**adls_options)
        .load(caminhos["silver_produtos"])
        .select("sku", "preco_lista").distinct())

    return df_silver_pedidos, df_silver_produtos

print("Função de carga de referências pronta.")
```

```python
def detectar_novo_bronze(nome_tabela, caminho_bronze):
    df_bronze = spark.read.format("delta").options(**adls_options).load(caminho_bronze)

    if tabela_delta_existe(caminhos["metadata_silver_control_log"]):
        df_control = spark.read.format("delta").options(**adls_options).load(caminhos["metadata_silver_control_log"])
        ja_processados = {
            r["arquivo_processado"] for r in
            df_control.filter(F.col("tabela_origem") == nome_tabela).select("arquivo_processado").collect()
        }
    else:
        ja_processados = set()

    df_novos = df_bronze.filter(~F.col("bronze_source_file").isin(ja_processados)) if ja_processados else df_bronze

    arquivos_no_lote = [r["bronze_source_file"] for r in df_novos.select("bronze_source_file").distinct().collect()]

    return df_novos, arquivos_no_lote

print("Função de detecção incremental (Bronze -> Silver) pronta.")
```

### Etapa 5 — Regras de qualidade: `ecommerce_itens_pedido`

Implementa as 8 regras por linha como colunas booleanas. As FKs (`id_pedido` e `sku`) são verificadas com `join` contra as Silvers de referência, sem trazer os IDs para o driver. A regra 2 fica **pendente** (`null`) quando a Silver de `ecommerce_pedidos` ainda não está disponível. As regras 7 e 10 são agregadas por pedido e ficam no notebook 06.

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

def aplicar_regras_itens_pedido(df_novos, df_silver_pedidos, df_silver_produtos, pedidos_disponivel):
    # Regra técnica 1: id_item_pedido não nulo e não duplicado dentro do micro-lote
    janela_dedup_item = Window.partitionBy("id_item_pedido")
    df = df_novos.withColumn(
        "tecnica_1_pk_valida",
        F.col("id_item_pedido").isNotNull() & (F.count("*").over(janela_dedup_item) == 1))

    # Regra técnica 2: id_pedido existe em ecommerce_pedidos (FK, via join)
    if pedidos_disponivel:
        ref_pedidos = df_silver_pedidos.select("id_pedido").distinct().withColumn("_pedido_existe", F.lit(True))
        df = (df.join(ref_pedidos, "id_pedido", "left")
                .withColumn("tecnica_2_fk_pedido_valido",
                            F.col("id_pedido").isNotNull() & F.col("_pedido_existe").isNotNull()))
    else:
        df = df.withColumn("tecnica_2_fk_pedido_valido", F.lit(None).cast("boolean"))  # pendente

    # Regra técnica 3: sku existe em ecommerce_produtos (FK, via join)
    ref_skus = df_silver_produtos.select("sku").distinct().withColumn("_sku_existe", F.lit(True))
    df = (df.join(ref_skus, "sku", "left")
            .withColumn("tecnica_3_fk_sku_valido", F.col("sku").isNotNull() & F.col("_sku_existe").isNotNull()))

    df_regras = (df
        # Regra técnica 4: quantidade >= 1 e inteira
        .withColumn("tecnica_4_quantidade_valida",
                    (F.col("quantidade") >= 1) & (F.col("quantidade") == F.floor(F.col("quantidade"))))
        # Regra técnica 5: preco_unitario > 0
        .withColumn("tecnica_5_preco_valido", F.col("preco_unitario") > 0)
        # Regra de negócio 6: desconto_aplicado não pode ser maior que preco_unitario
        .withColumn("negocio_6_desconto_nao_maior_preco", F.col("desconto_aplicado") <= F.col("preco_unitario"))
        # Regra de negócio 8: desconto_aplicado não pode ser negativo
        .withColumn("negocio_8_desconto_nao_negativo", F.col("desconto_aplicado") >= 0)
        # Regra de negócio 9: desconto até 50% do preco_unitario
        .withColumn("negocio_9_desconto_dentro_limite", F.col("desconto_aplicado") <= (F.col("preco_unitario") * 0.5))
        .drop("_pedido_existe", "_sku_existe"))

    # Regras de negócio 7 e 10 são agregadas por pedido: calculadas e gravadas no notebook 06.
    return df_regras

print("Função de aplicação de regras para itens_pedido pronta.")
```

### Etapa 6 — Regras de qualidade: `ecommerce_rastreamento_entregas`

Implementa as 6 regras por linha como colunas booleanas. A regra 6 (ordem cronológica) combina o micro-lote novo com o histórico já gravado na Silver **dos mesmos pedidos** (via `left_semi` join), para comparar corretamente a sequência de eventos. A FK da regra 3 é verificada com `join` e fica **pendente** (`null`) sem a Silver de `ecommerce_pedidos`. As regras 7 a 10 são agregadas por pedido e ficam no notebook 06.

```python
STATUS_VALIDOS = ["em separacao", "coletado", "em transito", "saiu para entrega", "entregue"]

def aplicar_regras_rastreamento(df_novos, df_silver_pedidos, df_silver_rastreamento_existente, pedidos_disponivel):
    # Regra 6: combina o lote novo com o histórico já gravado dos MESMOS pedidos
    if df_silver_rastreamento_existente is not None:
        pedidos_do_lote = df_novos.select("id_pedido_ecommerce").distinct()
        historico_relevante = (df_silver_rastreamento_existente
            .select("id_pedido_ecommerce", "dt_evento")
            .join(pedidos_do_lote, "id_pedido_ecommerce", "left_semi"))
    else:
        historico_relevante = None

    df_novos_marcado = df_novos.withColumn("_e_novo", F.lit(True))
    combinado = (df_novos_marcado.unionByName(historico_relevante.withColumn("_e_novo", F.lit(False)), allowMissingColumns=True)
                 if historico_relevante is not None else df_novos_marcado)

    janela_ordem = Window.partitionBy("id_pedido_ecommerce").orderBy("dt_evento")
    combinado = combinado.withColumn("_dt_evento_anterior", F.lag("dt_evento").over(janela_ordem))
    combinado = combinado.withColumn(
        "negocio_6_ordem_cronologica",
        F.col("_dt_evento_anterior").isNull() | (F.col("dt_evento") >= F.col("_dt_evento_anterior"))
    )

    # Regra técnica 1: id_rastreamento não nulo e não duplicado dentro do micro-lote
    janela_dedup = Window.partitionBy("id_rastreamento")
    df = (combinado.filter(F.col("_e_novo"))
        .drop("_e_novo", "_dt_evento_anterior")
        .withColumn("tecnica_1_pk_valida",
                    F.col("id_rastreamento").isNotNull() & (F.count("*").over(janela_dedup) == 1)))

    # Regra técnica 3: id_pedido_ecommerce existe em ecommerce_pedidos (FK, via join)
    if pedidos_disponivel:
        ref_pedidos = (df_silver_pedidos.select(F.col("id_pedido").alias("id_pedido_ecommerce")).distinct()
            .withColumn("_pedido_existe", F.lit(True)))
        df = (df.join(ref_pedidos, "id_pedido_ecommerce", "left")
                .withColumn("tecnica_3_fk_pedido_valido",
                            F.col("id_pedido_ecommerce").isNotNull() & F.col("_pedido_existe").isNotNull()))
    else:
        df = df.withColumn("tecnica_3_fk_pedido_valido", F.lit(None).cast("boolean"))  # pendente

    df_regras = (df
        # Regra técnica 2: status dentro do domínio
        .withColumn("tecnica_2_status_valido", F.col("status_entrega").isin(STATUS_VALIDOS))
        # Regra técnica 4: dt_evento preenchida e não futura
        .withColumn("tecnica_4_data_valida", F.col("dt_evento").isNotNull() & (F.col("dt_evento") <= F.current_timestamp()))
        # Regra técnica 5: código no formato TRK + 6 dígitos
        .withColumn("tecnica_5_codigo_rastreio_valido", F.col("codigo_rastreio").rlike(r"^TRK\d{6}$"))
        .drop("_pedido_existe"))

    # Regras de negócio 7 a 10 são agregadas por pedido: calculadas e gravadas no notebook 06.
    return df_regras

print("Função de aplicação de regras para rastreamento_entregas pronta.")
```

### Etapa 7 — Gravação na Silver + log de controle

Grava o resultado (com todas as colunas de regras) em modo `append` na Silver, adiciona `silver_processed_at`, e registra cada arquivo promovido no log de controle da Silver.


```python
from pyspark.sql.types import StructType, StructField, StringType, LongType
from datetime import datetime

schema_log = StructType([
    StructField("tabela_origem", StringType(), False),
    StructField("arquivo_processado", StringType(), False),
    StructField("total_linhas", LongType(), False),
    StructField("status", StringType(), False),
    StructField("timestamp_processamento", StringType(), False),
])

def processar_silver(df_regras, nome_tabela, caminho_silver):
    if df_regras is None:
        print(f"    Nenhum arquivo novo para '{nome_tabela}'.")
        return

    total = df_regras.count()
    resumo_por_arquivo = df_regras.groupBy("bronze_source_file").agg(F.count("*").alias("linhas")).collect()
    print(f"    >>> {len(resumo_por_arquivo)} arquivo(s) de origem promovido(s) para Silver de '{nome_tabela}' ({total} linhas)")

    df_final = df_regras.withColumn("silver_processed_at", F.current_timestamp())

    (df_final.write
        .format("delta")
        .options(**adls_options)
        .mode("append")
        .save(caminho_silver))

    print(f"    [OK] Silver de '{nome_tabela}' gravada.")

    try:
        agora = str(datetime.now())
        log_data = [(nome_tabela, r["bronze_source_file"], int(r["linhas"]), "SUCCESS", agora) for r in resumo_por_arquivo]
        df_log = spark.createDataFrame(log_data, schema=schema_log)

        (df_log.write
            .format("delta")
            .options(**adls_options)
            .mode("append")
            .save(caminhos["metadata_silver_control_log"]))

        print(f"    [OK] Log de controle Silver atualizado para '{nome_tabela}'.")
    except Exception as e_log:
        print(f"    [ERRO NO LOG] {type(e_log).__name__}: {e_log}")

print("Função de gravação na Silver pronta.")
```

### Etapa 8 — Loop de processamento

A cada verificação (15s, por até 90s), recarrega a disponibilidade da Silver de `pedidos` (pra promover regras de pendente pra ativas automaticamente quando a dependência aparecer), detecta e processa o que for novo nas duas tabelas, com a mesma proteção de 5 falhas consecutivas do notebook `04`.


```python
import time
from datetime import datetime, timedelta

intervalo_segundos = 15
duracao_maxima_segundos = 90
inicio = datetime.now()
fim_previsto = inicio + timedelta(seconds=duracao_maxima_segundos)
falhas_consecutivas = 0
MAX_FALHAS_CONSECUTIVAS = 5

print(f"Iniciando processamento Silver às {inicio.strftime('%H:%M:%S')}.")
print(f"Vai rodar até {fim_previsto.strftime('%H:%M:%S')} (ou até interromper manualmente).\n")

while datetime.now() < fim_previsto:
    try:
        print(f"--- Verificação às {datetime.now().strftime('%H:%M:%S')} ---")

        # Referências compartilhadas (recarregadas a cada verificação, pra pegar dado novo se existir)
        pedidos_disponivel = tabela_delta_existe(caminhos["silver_pedidos"])
        df_silver_pedidos = (spark.read.format("delta").options(**adls_options).load(caminhos["silver_pedidos"])
                              if pedidos_disponivel else None)
        df_silver_produtos = (spark.read.format("delta").options(**adls_options)
                               .load(caminhos["silver_produtos"]).select("sku").distinct())

        if not pedidos_disponivel:
            print("    [AVISO] Silver de pedidos ainda indisponível — regras dependentes ficarão pendentes.")

        # --- ecommerce_itens_pedido ---
        df_novos_itens, arquivos_itens = detectar_novo_bronze("ecommerce_itens_pedido", caminhos["bronze_itens_pedido"])
        if arquivos_itens:
            df_regras_itens = aplicar_regras_itens_pedido(df_novos_itens, df_silver_pedidos, df_silver_produtos, pedidos_disponivel)
            processar_silver(df_regras_itens, "ecommerce_itens_pedido", caminhos["silver_itens_pedido"])
        else:
            print("    Nenhum arquivo novo para 'ecommerce_itens_pedido'.")

        # --- ecommerce_rastreamento_entregas ---
        rastreamento_existe = tabela_delta_existe(caminhos["silver_rastreamento"])
        df_silver_rastreamento_existente = (spark.read.format("delta").options(**adls_options).load(caminhos["silver_rastreamento"])
                                             if rastreamento_existe else None)

        df_novos_rastreamento, arquivos_rastreamento = detectar_novo_bronze("ecommerce_rastreamento_entregas", caminhos["bronze_rastreamento"])
        if arquivos_rastreamento:
            df_regras_rastreamento = aplicar_regras_rastreamento(df_novos_rastreamento, df_silver_pedidos, df_silver_rastreamento_existente, pedidos_disponivel)
            processar_silver(df_regras_rastreamento, "ecommerce_rastreamento_entregas", caminhos["silver_rastreamento"])
        else:
            print("    Nenhum arquivo novo para 'ecommerce_rastreamento_entregas'.")

        falhas_consecutivas = 0

    except KeyboardInterrupt:
        print("\nProcessamento interrompido manualmente.")
        break
    except Exception as e:
        falhas_consecutivas += 1
        print(f"    [ERRO] Falha na verificação ({falhas_consecutivas}/{MAX_FALHAS_CONSECUTIVAS}): {e}")
        if falhas_consecutivas >= MAX_FALHAS_CONSECUTIVAS:
            print(f"    {MAX_FALHAS_CONSECUTIVAS} falhas seguidas — interrompendo o loop.")
            break

    time.sleep(intervalo_segundos)

print(f"\nProcessamento finalizado às {datetime.now().strftime('%H:%M:%S')}.")
```
