### Objetivo:
Gravação segura das credenciais no Databricks Secret Scope, evitando exposição em arquivos `.env`/notebook versionado no Git. Execução única — só roda de novo se alguma credencial precisar ser atualizada.


```python
dbutils.widgets.text("pat_token", "", "PAT (temporário)")
dbutils.widgets.text("secret_key", "", "Nome da chave (ex: adls-client-secret)")
dbutils.widgets.text("secret_value", "", "Valor do secret")
```

### Etapa 1 — Gravação do secret

Lê os valores preenchidos nos widgets acima e grava a chave/valor no Secret Scope via API REST do Databricks. O PAT é usado só nesta chamada, nunca fica salvo em lugar nenhum do notebook.


```python
import requests

workspace_url = "https://dbc-bdcb97c0-9299.cloud.databricks.com"
scope = "internship-squad1"

pat = dbutils.widgets.get("pat_token")
key = dbutils.widgets.get("secret_key")
value = dbutils.widgets.get("secret_value")

resp = requests.post(
    f"{workspace_url}/api/2.0/secrets/put",
    headers={"Authorization": f"Bearer {pat}"},
    json={"scope": scope, "key": key, "string_value": value},
)

print(f"Status: {resp.status_code} | chave '{key}' gravada: {resp.ok}")
```
