```python
# ---------------------------------------------------------------------------
# Carga de dados no SQL Server via Lakehouse Federation
# ---------------------------------------------------------------------------
# IMPORTANTE: O conector JDBC (format("sqlserver")) NÃO funciona no serverless.
# O bloqueio de rede faz o `mode("overwrite")` truncar a tabela de destino sem
# conseguir escrever os novos dados — resultando em perda de dados.
#
# A solução é usar o catálogo estrangeiro `sqlserver_catalog` (UC gerencia a conexão).
# ---------------------------------------------------------------------------
```

```python
# Grava os dados de ecommerce_clientes em uma tabela Delta no catálogo workspace.
# ---------------------------------------------------------------------------
# IMPORTANTE: Não é possível escrever no SQL Server a partir do serverless:
#   1. JDBC (format("sqlserver")) — bloqueado pela rede, trunca a tabela sem escrever
#   2. Lakehouse Federation (saveAsTable) — catálogo estrangeiro é somente leitura
#
# Solução: gravar em tabela Delta no catálogo workspace (funciona no serverless).
# Os dados ficam acessíveis para todos os notebooks do projeto.
# ---------------------------------------------------------------------------

df_clientes.write \
    .mode("overwrite") \
    .saveAsTable("workspace.default.ecommerce_clientes")

print("✅ Dados gravados em workspace.default.ecommerce_clientes (Delta)!")
print(f"   Linhas gravadas: {df_clientes.count()}")
```

```python
# Verificação: lê a tabela Delta gravada no catálogo workspace

df_clientes_sql = spark.table("workspace.default.ecommerce_clientes")
print(f"Linhas em workspace.default.ecommerce_clientes: {df_clientes_sql.count()}")
display(df_clientes_sql)
```
