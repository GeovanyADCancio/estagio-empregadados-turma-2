## Tabelas analisadas
- **E-commerce:** categorias, clientes, endereços, itens_pedido, pedidos, produtos, rastreamento_entregas  
- **Food:** avaliações_produto, estoque_lojas, fornecedores, lotes_producao  
- **Physical:** itens_venda_caixa, lojas, produtos_pereciveis, vendas_caixa

```python
from pyspark.sql import functions as F

def analisar_tabela(nome_tabela):
    print(f"\n{'='*60}")
    print(f"📋 {nome_tabela}")
    print('='*60)
    
    df = spark.table(f"workspace.squad3.{nome_tabela}")
    
    # Shape
    linhas = df.count()
    colunas = len(df.columns)
    print(f"📐 Shape: {linhas:,} linhas × {colunas} colunas")
    
    # Schema
    print("\n📌 Schema:")
    df.printSchema()
    
    # Nulos por coluna
    print("🔍 Valores nulos por coluna:")
    nulos = df.select([
        F.sum(F.col(c).isNull().cast("int")).alias(c)
        for c in df.columns
    ])
    nulos.show(truncate=False)
    
    # Duplicatas
    total = df.count()
    distintos = df.dropDuplicates().count()
    duplicatas = total - distintos
    print(f"🔁 Registros duplicados: {duplicatas:,}")
    
    # Estatísticas descritivas
    print("📊 Estatísticas descritivas:")
    df.describe().show(truncate=False)
    
    return df
```

## 1️⃣ E-commerce

```python
tabelas_ecommerce = [
    "ecommerce_categorias",
    "ecommerce_clientes",
    "ecommerce_enderecos",
    "ecommerce_itens_pedido",
    "ecommerce_pedidos",
    "ecommerce_produtos",
    "ecommerce_rastreamento_entregas",
]

for tabela in tabelas_ecommerce:
    analisar_tabela(tabela)
```

## 2️⃣ Food

```python
tabelas_food = [
    "food_avaliacoes_produto",
    "food_estoque_lojas",
    "food_fornecedores",
    "food_lotes_producao",
]

for tabela in tabelas_food:
    analisar_tabela(tabela)
```

## 3️⃣ Physical

```python
tabelas_physical = [
    "physical_itens_venda_caixa",
    "physical_lojas",
    "physical_produtos_pereciveis",
    "physical_vendas_caixa",
]

for tabela in tabelas_physical:
    analisar_tabela(tabela)
```

## ✅ Resumo Geral

```python
todas_tabelas = [
    "ecommerce_categorias", "ecommerce_clientes", "ecommerce_enderecos",
    "ecommerce_itens_pedido", "ecommerce_pedidos", "ecommerce_produtos",
    "ecommerce_rastreamento_entregas", "food_avaliacoes_produto",
    "food_estoque_lojas", "food_fornecedores", "food_lotes_producao",
    "physical_itens_venda_caixa", "physical_lojas",
    "physical_produtos_pereciveis", "physical_vendas_caixa",
]

print(f"{'Tabela':<40} {'Linhas':>12} {'Colunas':>8}")
print("-" * 62)

total_linhas = 0
for tabela in todas_tabelas:
    df = spark.table(f"workspace.squad3.{tabela}")
    linhas = df.count()
    colunas = len(df.columns)
    total_linhas += linhas
    print(f"{tabela:<40} {linhas:>12,} {colunas:>8}")

print("-" * 62)
print(f"{'TOTAL':<40} {total_linhas:>12,}")
```
