```python


# Teste de Permissão - Spark JDBC Nativo
import os
from dotenv import load_dotenv

# 1. Carrega as variáveis do arquivo .env
load_dotenv()

SQL_HOST     = os.getenv("srv-database-intership.database.windows.net")
SQL_DATABASE = os.getenv("internshipDatabase")
SQL_USERNAME = os.getenv(estagiario_user")
SQL_PASSWORD = os.getenv("InternshipAnalytics2026TurmaSetembro")

# Validação das variáveis
if not all([SQL_HOST, SQL_DATABASE, SQL_USERNAME, SQL_PASSWORD]):
    raise ValueError("❌ Erro: Credenciais do SQL não encontradas no arquivo .env!")

print(f"Iniciando teste de conexão via Conector Nativo SQL Server em: {SQL_HOST}...")

# No formato nativo 'sqlserver', passamos os parâmetros de conexão diretamente nas opções
tabela_teste = "gold_teste_permissao_spark"

try:
    # 2. Criar dados fictícios no Spark
    dados_teste = [("Databricks Serverless", "Sucesso")]
    colunas = ["origem", "status"]
    df_teste = spark.createDataFrame(dados_teste, colunas)
    
    # 3. Executar a Escrita mudando o formato para 'sqlserver'
    print("\n[Teste] Validando permissão de escrita via formato sqlserver...")
    df_teste.write \
        .format("sqlserver") \
        .option("host", SQL_HOST) \
        .option("port", "1433") \
        .option("database", SQL_DATABASE) \
        .option("user", SQL_USERNAME) \
        .option("password", SQL_PASSWORD) \
        .option("dbtable", tabela_teste) \
        .mode("overwrite") \
        .save()
        
    print(f"  -> Tabela '{tabela_teste}' gravada com sucesso na Gold!")

    # 4. Executar a Leitura para validação final
    df_validacao = spark.read \
        .format("sqlserver") \
        .option("host", SQL_HOST) \
        .option("port", "1433") \
        .option("database", SQL_DATABASE) \
        .option("user", SQL_USERNAME) \
        .option("password", SQL_PASSWORD) \
        .option("dbtable", tabela_teste) \
        .load()
        
    print(f"\n🎉 SUCESSO! Dados gravados e lidos no Databricks Serverless:")
    display(df_validacao)

except Exception as e:
    print("\n❌ ERRO DE CONEXÃO OU PERMISSÃO NO CONECTOR:")
    print(f"Detalhes: {e}")

     
```
