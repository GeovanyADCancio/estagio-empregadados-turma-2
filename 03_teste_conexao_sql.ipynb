## 🔌 Teste de Conexão — Azure SQL Database · Squad 3

Valida permissão de leitura e escrita (CREATE/INSERT) no Azure SQL Server.

```python
from dotenv import load_dotenv
import os

load_dotenv("/Workspace/Users/alexandrofs31@gmail.com/.env")

SQL_HOST     = os.getenv("SQL_HOST")
SQL_DATABASE = os.getenv("SQL_DATABASE")
SQL_USERNAME = os.getenv("SQL_USERNAME")
SQL_PASSWORD = os.getenv("SQL_PASSWORD")

jdbc_url = f"jdbc:sqlserver://{SQL_HOST}:1433;databaseName={SQL_DATABASE};encrypt=true;trustServerCertificate=false;loginTimeout=30"

print(f"🔗 URL: {jdbc_url}")
print(f"👤 Usuário: {SQL_USERNAME}")
```

```python
# Teste de Leitura

try:
    df_test = spark.read \
        .format("jdbc") \
        .option("url", jdbc_url) \
        .option("query", "SELECT 1 AS teste") \
        .option("user", SQL_USERNAME) \
        .option("password", SQL_PASSWORD) \
        .option("driver", "com.microsoft.sqlserver.jdbc.SQLServerDriver") \
        .load()
    
    df_test.show()
    print("✅ Conexão com SQL Server estabelecida com sucesso!")

except Exception as e:
    print(f"❌ Falha na conexão: {str(e)[:300]}")
```

```python
# Testar escrita (CREATE/INSERT)

from pyspark.sql import Row

try:
    df_write_test = spark.createDataFrame([
        Row(squad="squad3", autor="alexandro", status="ok")
    ])

    df_write_test.write \
        .format("sqlserver") \
        .option("host", SQL_HOST) \
        .option("port", "1433") \
        .option("database", SQL_DATABASE) \
        .option("dbtable", "squad3.teste_conexao_alexandro") \
        .option("user", SQL_USERNAME) \
        .option("password", SQL_PASSWORD) \
        .option("encrypt", "true") \
        .option("trustServerCertificate", "false") \
        .mode("overwrite") \
        .save()

    print("✅ Escrita no SQL Server confirmada — permissão CREATE/INSERT OK!")

except Exception as e:
    print(f"❌ Falha na escrita: {str(e)[:300]}")
```
