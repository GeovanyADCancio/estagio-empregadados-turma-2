```python
# ============================================================================
# Extração de dados via Unity Catalog Lakehouse Federation
# ----------------------------------------------------------------------------
# O serverless bloqueia conexões diretas (Azure SDK, JDBC). A solução é ler
# via catálogo estrangeiro `sqlserver_catalog` (conexão gerenciada pelo UC).
#
# IMPORTANTE: Algumas tabelas dbo foram esvaziadas por JDBC mode("overwrite")
# no serverless. Por isso lemos de cada schema onde os dados ainda existem:
#   - Produtos:  squad2 (dbo está vazia)
#   - Categorias: dbo    (1.620 linhas)
#   - Clientes:  dbo    (7.656 linhas)
#   - Pedidos:   squad2 (90.335 linhas, schema completo com dt_previsao_entrega)
# ============================================================================

# Leitura direta via Spark SQL — funciona no serverless!
df_produtos = spark.table("sqlserver_catalog.squad2.ecommerce_produtos")
df_categorias = spark.table("sqlserver_catalog.dbo.ecommerce_categorias")
df_clientes = spark.table("sqlserver_catalog.dbo.ecommerce_clientes")
df_pedidos = spark.table("sqlserver_catalog.squad2.ecommerce_pedidos")

# Validação final e exibição
print(f"🚀 Leitura via Lakehouse Federation concluída!")
print(f"   Produtos:  {df_produtos.count()} linhas")
print(f"   Categorias: {df_categorias.count()} linhas")
print(f"   Clientes:  {df_clientes.count()} linhas")
print(f"   Pedidos:   {df_pedidos.count()} linhas")
print()
display(df_produtos.limit(5))
```
